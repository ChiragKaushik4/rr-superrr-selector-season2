```python
import pandas as pd
import numpy as np

# ---------------------------------------------------------
# PHASE 2: COMPETITION SCOPE
# Include both IPL and T20/T20I competitions as per the brief
# ---------------------------------------------------------
t20_matches = matches[
    matches['comp_name'].str.contains('IPL|T20', case=False, na=False)
].copy()

t20_ids = t20_matches['match_id'].unique()

df_ball = ball[
    (ball['match_id'].isin(t20_ids)) &
    (ball['innings'].isin([1, 2]))
].copy()

# ---------------------------------------------------------
# PHASE 3: MATCH-STATE CONSTRUCTION
# ---------------------------------------------------------
# 1. Clean Extras & Define Legal Balls
df_ball['extras1_runs'] = pd.to_numeric(df_ball['extras1_runs'], errors='coerce').fillna(6)
df_ball['is_legal_ball'] = ~df_ball['extras1_runs'].isin([1, 2, 5])

# 2. Fix the Missing Extras Run Problem
# Assign a standard 1-run penalty to Wides (2) and No-balls (1).
df_ball['extra_penalty_runs'] = np.where(df_ball['extras1_runs'].isin([1, 2]), 1, 0)
df_ball['delivery_total_runs'] = df_ball['runs'] + df_ball['extra_penalty_runs']

# 3. Calculate Cumulative Match States
# Sort chronologically to ensure rolling sums work perfectly
df_ball = df_ball.sort_values(by=['match_id', 'innings', 'overid', 'over_ball_num'])

df_ball['cumulative_runs'] = df_ball.groupby(['match_id', 'innings'])['delivery_total_runs'].cumsum()
df_ball['cumulative_wickets'] = df_ball.groupby(['match_id', 'innings'])['is_wicket'].cumsum()
df_ball['legal_balls_bowled'] = df_ball.groupby(['match_id', 'innings'])['is_legal_ball'].cumsum()

# Balls remaining (Standard T20 is 120 balls)
df_ball['balls_remaining'] = 120 - df_ball['legal_balls_bowled']
df_ball['balls_remaining'] = df_ball['balls_remaining'].clip(lower=0)

# 4. Extract Target Scores for Second Innings
first_innings_score = df_ball[df_ball['innings'] == 1].groupby('match_id')['cumulative_runs'].max().reset_index()
first_innings_score.rename(columns={'cumulative_runs': 'target_runs'}, inplace=True)
first_innings_score['target_runs'] += 1 # Target is 1st innings score + 1

df_ball = df_ball.merge(first_innings_score, on='match_id', how='left')

# 5. Calculate Required Run Rate (RRR) for Second Innings
df_ball['runs_required'] = np.where(
    df_ball['innings'] == 2,
    df_ball['target_runs'] - df_ball['cumulative_runs'],
    np.nan
)

# Avoid division by zero on the final ball
df_ball['rrr'] = np.where(
    (df_ball['innings'] == 2) & (df_ball['balls_remaining'] > 0),
    (df_ball['runs_required'] / df_ball['balls_remaining']) * 6,
    np.nan
)

print(f"Total T20 Matches in Scope: {len(t20_ids)}")
print(f"Total Deliveries Processed: {len(df_ball)}")

# Display a snapshot of a run chase to verify the rolling math
chase_snapshot = df_ball[
    (df_ball['match_id'] == t20_ids[0]) &
    (df_ball['innings'] == 2)
][['overid', 'over_ball_num', 'delivery_total_runs', 'cumulative_runs', 'cumulative_wickets', 'balls_remaining', 'runs_required', 'rrr']].tail(10)

print("\n--- End of Innings 2 Chase Snapshot ---")
display(chase_snapshot)
```

```python
import pandas as pd
import numpy as np

# ---------------------------------------------------------
# PHASE 4: PRESSURE CLASSIFICATION MODEL
# ---------------------------------------------------------

# --- FIRST INNINGS PRESSURE CONDITIONS ---
# 1. Early Collapse: 3 or more wickets down inside the Powerplay (first 36 balls)
cond_1st_collapse = (df_ball['innings'] == 1) & (df_ball['cumulative_wickets'] >= 3) & (df_ball['legal_balls_bowled'] <= 36)

# 2. Mid-Innings Rescue: 5 or more wickets down before the 15th over (rebuilding phase)
cond_1st_rescue = (df_ball['innings'] == 1) & (df_ball['cumulative_wickets'] >= 5) & (df_ball['legal_balls_bowled'] <= 90)

# 3. Death Overs Expectation: The final 4 overs (24 balls) are inherently high pressure to maximize the total
cond_1st_death = (df_ball['innings'] == 1) & (df_ball['balls_remaining'] <= 24)

# Combine 1st Innings Pressure
pressure_1st = cond_1st_collapse | cond_1st_rescue | cond_1st_death


# --- SECOND INNINGS PRESSURE CONDITIONS ---
# 1. Steep Chase: Required Run Rate climbs above 10.5 at any point in the innings
cond_2nd_steep = (df_ball['innings'] == 2) & (df_ball['rrr'] >= 10.5)

# 2. Tight Death Chase: Last 5 overs (30 balls) needing 8.0 RRR or more
cond_2nd_death = (df_ball['innings'] == 2) & (df_ball['balls_remaining'] <= 30) & (df_ball['rrr'] >= 8.0) & (df_ball['runs_required'] > 0)

# 3. Chase Collapse: Lost 4 or more wickets but still need 50+ runs to win
cond_2nd_collapse = (df_ball['innings'] == 2) & (df_ball['cumulative_wickets'] >= 4) & (df_ball['runs_required'] >= 50)

# Combine 2nd Innings Pressure
pressure_2nd = cond_2nd_steep | cond_2nd_death | cond_2nd_collapse

# --- APPLY PRESSURE FLAG ---
df_ball['is_pressure'] = pressure_1st | pressure_2nd

# ---------------------------------------------------------
# VALIDATE THE PRESSURE MODEL
# ---------------------------------------------------------
# Let's see what percentage of the game we have defined as "Pressure"
total_balls = len(df_ball)
pressure_balls = df_ball['is_pressure'].sum()
pressure_pct = (pressure_balls / total_balls) * 100

print(f"Total Deliveries: {total_balls}")
print(f"Total Pressure Deliveries: {pressure_balls}")
print(f"Percentage of match played under pressure: {pressure_pct:.2f}%\n")

# Breakdown by Innings
pressure_breakdown = df_ball.groupby('innings').agg(
    total_balls=('over_ball_num', 'count'),
    pressure_balls=('is_pressure', 'sum')
)
pressure_breakdown['pressure_pct'] = (pressure_breakdown['pressure_balls'] / pressure_breakdown['total_balls']) * 100

print("--- Pressure Breakdown by Innings ---")
print(pressure_breakdown)
```

```python
import pandas as pd
import numpy as np

# ---------------------------------------------------------
# PHASE 5: PLAYER PERFORMANCE METRICS
# ---------------------------------------------------------

# --- A. BATTER METRICS ---
# Aggregate stats by striker and pressure flag
batter_agg = df_ball.groupby(['striker_name', 'is_pressure']).agg(
    total_runs=('runs', 'sum'),
    legal_balls_faced=('is_legal_ball', 'sum'),
    boundaries=('runs', lambda x: (x.isin([4, 6])).sum()),
    dismissals=('is_wicket', 'sum') # Note: Simplification assuming striker is out
).reset_index()

# Pivot to compare Normal (False) vs Pressure (True)
batter_pivot = batter_agg.pivot(index='striker_name', columns='is_pressure', values=['total_runs', 'legal_balls_faced', 'boundaries', 'dismissals']).fillna(0)
batter_pivot.columns = ['_'.join(map(str, col)).strip() for col in batter_pivot.columns.values]
batter_pivot = batter_pivot.reset_index()

# Rename columns for clarity
batter_pivot.rename(columns={
    'total_runs_False': 'normal_runs', 'total_runs_True': 'pressure_runs',
    'legal_balls_faced_False': 'normal_balls', 'legal_balls_faced_True': 'pressure_balls',
    'boundaries_False': 'normal_boundaries', 'boundaries_True': 'pressure_boundaries',
    'dismissals_False': 'normal_dismissals', 'dismissals_True': 'pressure_dismissals'
}, inplace=True)

# Calculate Strike Rates (SR)
batter_pivot['normal_sr'] = np.where(batter_pivot['normal_balls'] > 0, (batter_pivot['normal_runs'] / batter_pivot['normal_balls']) * 100, 0)
batter_pivot['pressure_sr'] = np.where(batter_pivot['pressure_balls'] > 0, (batter_pivot['pressure_runs'] / batter_pivot['pressure_balls']) * 100, 0)

# Calculate Pressure SR Differential (Positive means they bat FASTER under pressure)
batter_pivot['sr_differential'] = batter_pivot['pressure_sr'] - batter_pivot['normal_sr']

# Filter for Sample Size: Minimum 60 pressure balls faced (10 overs)
valid_batters = batter_pivot[batter_pivot['pressure_balls'] >= 60].copy()


# --- B. BOWLER METRICS ---
# Aggregate stats by bowler and pressure flag
bowler_agg = df_ball.groupby(['bowler_name', 'is_pressure']).agg(
    total_runs_conceded=('delivery_total_runs', 'sum'),
    legal_balls_bowled=('is_legal_ball', 'sum'),
    wickets_taken=('is_wicket', 'sum')
).reset_index()

# Pivot for Bowlers
bowler_pivot = bowler_agg.pivot(index='bowler_name', columns='is_pressure', values=['total_runs_conceded', 'legal_balls_bowled', 'wickets_taken']).fillna(0)
bowler_pivot.columns = ['_'.join(map(str, col)).strip() for col in bowler_pivot.columns.values]
bowler_pivot = bowler_pivot.reset_index()

# Rename columns
bowler_pivot.rename(columns={
    'total_runs_conceded_False': 'normal_runs_conceded', 'total_runs_conceded_True': 'pressure_runs_conceded',
    'legal_balls_bowled_False': 'normal_balls_bowled', 'legal_balls_bowled_True': 'pressure_balls_bowled',
    'wickets_taken_False': 'normal_wickets', 'wickets_taken_True': 'pressure_wickets'
}, inplace=True)

# Calculate Economy Rates (Runs per 6 balls)
bowler_pivot['normal_econ'] = np.where(bowler_pivot['normal_balls_bowled'] > 0, (bowler_pivot['normal_runs_conceded'] / bowler_pivot['normal_balls_bowled']) * 6, 0)
bowler_pivot['pressure_econ'] = np.where(bowler_pivot['pressure_balls_bowled'] > 0, (bowler_pivot['pressure_runs_conceded'] / bowler_pivot['pressure_balls_bowled']) * 6, 0)

# Calculate Pressure Economy Differential (Positive means they bowl TIGHTER/BETTER under pressure)
bowler_pivot['econ_differential'] = bowler_pivot['normal_econ'] - bowler_pivot['pressure_econ']

# Filter for Sample Size: Minimum 60 pressure balls bowled (10 overs)
valid_bowlers = bowler_pivot[bowler_pivot['pressure_balls_bowled'] >= 60].copy()

# --- QUICK PREVIEW ---
print(f"Total Batters with 60+ pressure balls: {len(valid_batters)}")
print("\n--- Top 5 Batters by Pressure Strike Rate ---")
display(valid_batters[['striker_name', 'pressure_runs', 'pressure_balls', 'pressure_sr', 'normal_sr', 'sr_differential']].sort_values(by='pressure_sr', ascending=False).head(5))

print(f"\nTotal Bowlers with 60+ pressure balls: {len(valid_bowlers)}")
print("\n--- Top 5 Bowlers by Pressure Economy (Lowest is Best) ---")
display(valid_bowlers[['bowler_name', 'pressure_runs_conceded', 'pressure_balls_bowled', 'pressure_econ', 'normal_econ', 'econ_differential']].sort_values(by='pressure_econ', ascending=True).head(5))
```

```python
import pandas as pd
import numpy as np
from scipy.stats import zscore
from sklearn.preprocessing import MinMaxScaler

# ---------------------------------------------------------
# PHASE 6: STATISTICAL FAIRNESS AND RANKING (PPI)
# ---------------------------------------------------------

# 1. Stricter Baseline Filter (Minimum 120 balls = 20 overs of pure pressure)
batters = batter_pivot[batter_pivot['pressure_balls'] >= 120].copy()
bowlers = bowler_pivot[bowler_pivot['pressure_balls_bowled'] >= 120].copy()

# --- A. BATTER BAYESIAN SHRINKAGE & Z-SCORES ---
# Global means for batters
global_mean_sr = batters['pressure_runs'].sum() / batters['pressure_balls'].sum() * 100
C_batter = 120 # Shrinkage weight (equivalent to 120 balls)

# Bayesian Shrunk Strike Rate
batters['shrunk_pressure_sr'] = (
    (batters['pressure_balls'] * batters['pressure_sr']) + (C_batter * global_mean_sr)
) / (batters['pressure_balls'] + C_batter)

# Z-Scores for the 3 pillars
batters['z_sr'] = zscore(batters['shrunk_pressure_sr'])
batters['z_diff'] = zscore(batters['sr_differential'])
batters['z_vol'] = zscore(batters['pressure_runs'])

# Composite Score (50% Performance, 25% Elevation, 25% Volume)
batters['raw_ppi'] = (batters['z_sr'] * 0.50) + (batters['z_diff'] * 0.25) + (batters['z_vol'] * 0.25)

# Normalize to a clean 0-100 index
scaler = MinMaxScaler(feature_range=(0, 100))
batters['Pressure_Performance_Index'] = scaler.fit_transform(batters[['raw_ppi']])


# --- B. BOWLER BAYESIAN SHRINKAGE & Z-SCORES ---
# Global means for bowlers
global_mean_econ = (bowlers['pressure_runs_conceded'].sum() / bowlers['pressure_balls_bowled'].sum()) * 6
C_bowler = 120

# Bayesian Shrunk Economy
bowlers['shrunk_pressure_econ'] = (
    (bowlers['pressure_balls_bowled'] * bowlers['pressure_econ']) + (C_bowler * global_mean_econ)
) / (bowlers['pressure_balls_bowled'] + C_bowler)

# Z-Scores for the 3 pillars
# Note: For Economy, LOWER is better, so we invert the Z-score (multiply by -1)
bowlers['z_econ'] = zscore(bowlers['shrunk_pressure_econ']) * -1
bowlers['z_diff_bowl'] = zscore(bowlers['econ_differential'])
bowlers['z_vol_bowl'] = zscore(bowlers['pressure_wickets'])

# Composite Score (50% Performance, 25% Elevation, 25% Volume)
bowlers['raw_ppi'] = (bowlers['z_econ'] * 0.50) + (bowlers['z_diff_bowl'] * 0.25) + (bowlers['z_vol_bowl'] * 0.25)

# Normalize to a clean 0-100 index
bowlers['Pressure_Performance_Index'] = scaler.fit_transform(bowlers[['raw_ppi']])

# ---------------------------------------------------------
# FINAL OUTPUT FORMATTING
# ---------------------------------------------------------
final_batters = batters.sort_values(by='Pressure_Performance_Index', ascending=False).reset_index(drop=True)
final_batters.index += 1 # 1-based ranking

final_bowlers = bowlers.sort_values(by='Pressure_Performance_Index', ascending=False).reset_index(drop=True)
final_bowlers.index += 1

print(f"Global T20 Pressure Strike Rate Baseline: {global_mean_sr:.2f}")
print(f"Global T20 Pressure Economy Baseline: {global_mean_econ:.2f}\n")

print("========== TOP 10 BATTERS UNDER PRESSURE ==========")
display_batters = final_batters[['striker_name', 'Pressure_Performance_Index', 'pressure_runs', 'pressure_sr', 'sr_differential']].head(10)
display_batters['Pressure_Performance_Index'] = display_batters['Pressure_Performance_Index'].round(2)
display(display_batters)

print("\n========== TOP 10 BOWLERS UNDER PRESSURE ==========")
display_bowlers = final_bowlers[['bowler_name', 'Pressure_Performance_Index', 'pressure_wickets', 'pressure_econ', 'econ_differential']].head(10)
display_bowlers['Pressure_Performance_Index'] = display_bowlers['Pressure_Performance_Index'].round(2)
display(display_bowlers)
```

```python
import os
import pandas as pd
import numpy as np
from scipy.stats import zscore

# 1. Load datasets (auto-detects /content/ or root / path)
base = "/" if os.path.exists("/ball_by_ball.csv") else ""
ball = pd.read_csv(f"{base}ball_by_ball.csv")
matches = pd.read_csv(f"{base}matches.csv")
competitions = pd.read_csv(f"{base}competitions.csv")

# 2. Merge competition metadata & filter to senior IPL + T20I (2022–2026)
matches_full = matches.merge(
    competitions[['comp_id', 'comp_name', 'season', 'match_format']], on='comp_id', how='left'
)
df = ball[ball['innings'].isin([1, 2])].merge(
    matches_full[['match_id', 'comp_id', 'comp_name', 'season', 'match_format', 'match_date',
                  'first_inn_revised_overs', 'second_inn_revised_overs', 'revised_target']],
    on='match_id', how='inner'
)
df['comp_tier'] = np.where(
    df['comp_name'].str.contains('IPL', case=False, na=False), 'IPL',
    np.where(df['match_format'] == 'T20I', 'T20I', 'Other_T20')
)
df = df[df['comp_tier'].isin(['IPL', 'T20I'])].copy()
df['season_clean'] = df['season'].astype(str).str[:4]

# 3. Delivery & Dismissal Cleaning
df['extras1_runs'] = pd.to_numeric(df['extras1_runs'], errors='coerce').fillna(6).astype(int)
df['is_legal_ball'] = ~df['extras1_runs'].isin([1, 2, 5])
df['extra_penalty'] = np.where(df['extras1_runs'].isin([1, 2]), 1, 0) + df['extras_2_runs'].fillna(0)
df['delivery_total_runs'] = df['runs'] + df['extra_penalty']
df['batter_runs'] = np.where(df['extras1_runs'].isin([2, 3, 4]), 0, df['runs'])
df['is_batter_ball'] = ~df['extras1_runs'].isin([2, 5])
df['bowler_runs'] = np.where(
    df['extras1_runs'].isin([3, 4]), 0,
    df['runs'] + np.where(df['extras1_runs'].isin([1, 2]), 1, 0) + df['extras_2_runs'].fillna(0)
)
non_bowler_outs = ['RunOut', 'Retire Out', 'Obstracting the field', 'Obstructing the field']
df['is_bowler_wicket'] = df['is_wicket'] & (~df['dismissal_type'].isin(non_bowler_outs))

# 4. Pre-Delivery Match State (Zero Data Leakage)
df = df.sort_values(['match_id', 'innings', 'overid', 'over_ball_num', 'innings_ball_num']).reset_index(drop=True)
df['runs_before'] = df.groupby(['match_id', 'innings'])['delivery_total_runs'].cumsum().shift(1, fill_value=0)
df['wkts_before'] = df.groupby(['match_id', 'innings'])['is_wicket'].cumsum().shift(1, fill_value=0)
df['balls_bowled_before'] = df.groupby(['match_id', 'innings'])['is_legal_ball'].cumsum().shift(1, fill_value=0)

first_ball = df.groupby(['match_id', 'innings']).cumcount() == 0
df.loc[first_ball, ['runs_before', 'wkts_before', 'balls_bowled_before']] = 0

df['max_balls'] = np.where(df['innings'] == 1, df['first_inn_revised_overs'].fillna(120), df['second_inn_revised_overs'].fillna(120))
df['balls_remaining_before'] = (df['max_balls'] - df['balls_bowled_before']).clip(lower=1)

inn1_totals = df[df['innings'] == 1].groupby('match_id')['delivery_total_runs'].sum().reset_index().rename(columns={'delivery_total_runs': 'inn1_total'})
df = df.merge(inn1_totals, on='match_id', how='left')
df['target'] = np.where(df['innings'] == 2, np.where(df['revised_target'].notna(), df['revised_target'], df['inn1_total'] + 1), np.nan)
df['runs_required_before'] = np.where(df['innings'] == 2, df['target'] - df['runs_before'], np.nan)
df['rrr_before'] = np.where(
    (df['innings'] == 2) & (df['runs_required_before'] > 0),
    (df['runs_required_before'] / df['balls_remaining_before']) * 6.0,
    np.where(df['innings'] == 2, 0.0, np.nan)
)
df['crr_before'] = np.where(df['balls_bowled_before'] > 0, (df['runs_before'] / df['balls_bowled_before']) * 6.0, 0.0)
df['phase'] = np.where(df['balls_bowled_before'] < 36, 'Powerplay', np.where(df['balls_bowled_before'] < 90, 'Middle', 'Death'))

# 5. Pre-Delivery Pressure Classification
p1 = (df['innings'] == 1) & (
    ((df['wkts_before'] >= 3) & (df['balls_bowled_before'] < 36)) |
    ((df['balls_bowled_before'] >= 36) & (df['balls_bowled_before'] < 90) & ((df['wkts_before'] >= 4) | ((df['wkts_before'] >= 3) & (df['crr_before'] < 6.2)))) |
    (df['balls_remaining_before'] <= 24)
)
p2 = (df['innings'] == 2) & (
    ((df['runs_required_before'] > 0) & (((df['rrr_before'] >= 11.0) & (df['runs_required_before'] >= 15)) | ((df['balls_bowled_before'] >= 36) & (df['rrr_before'] >= 10.2) & (df['runs_required_before'] >= 20)))) |
    ((df['balls_remaining_before'] <= 30) & (df['rrr_before'] >= 8.5) & (df['runs_required_before'] > 0)) |
    (((df['wkts_before'] >= 4) & (df['runs_required_before'] >= 45)) | ((df['wkts_before'] >= 6) & (df['runs_required_before'] >= 15)))
)
df['is_pressure'] = p1 | p2

# 6. Context-Adjusted Expected Runs (xRuns)
bat_par = df[df['is_batter_ball']].groupby(['comp_tier', 'season_clean', 'innings', 'phase', 'is_pressure'])['batter_runs'].mean().reset_index().rename(columns={'batter_runs': 'exp_bat'})
bowl_par = df[df['is_legal_ball']].groupby(['comp_tier', 'season_clean', 'innings', 'phase', 'is_pressure'])['bowler_runs'].mean().reset_index().rename(columns={'bowler_runs': 'exp_bowl'})
df = df.merge(bat_par, on=['comp_tier', 'season_clean', 'innings', 'phase', 'is_pressure'], how='left').merge(bowl_par, on=['comp_tier', 'season_clean', 'innings', 'phase', 'is_pressure'], how='left')

df['batter_rae'] = np.where(df['is_batter_ball'], df['batter_runs'] - df['exp_bat'], 0.0)
df['bowler_rse'] = np.where(df['is_legal_ball'], df['exp_bowl'] - df['bowler_runs'], 0.0)
df['is_boundary'] = df['batter_runs'].isin([4, 6])
df['is_bat_dot'] = df['is_batter_ball'] & (df['batter_runs'] == 0)
df['is_bowl_dot'] = df['is_legal_ball'] & (df['bowler_runs'] == 0)

# 7. Batter Aggregation & Empirical Bayes Shrinkage
last_bat = df.groupby('striker_id')['match_date'].max().to_dict()
ipl_bat_p = df[(df['comp_tier'] == 'IPL') & df['is_pressure'] & df['is_batter_ball']].groupby('striker_id').size().to_dict()

outs_df = df[df['is_wicket'] & df['dismissed_player_id'].notna() & (df['dismissal_type'] != 'Retire Out')].copy()
outs_df['dismissed_player_id'] = outs_df['dismissed_player_id'].astype(int)
bat_outs = outs_df.groupby(['dismissed_player_id', 'is_pressure']).size().unstack(fill_value=0).reset_index()
bat_outs.columns = ['striker_id', 'outs_N', 'outs_P']

bat_agg = df[df['is_batter_ball']].groupby(['striker_id', 'striker_name', 'is_pressure']).agg(
    runs=('batter_runs', 'sum'), balls=('is_batter_ball', 'sum'), rae=('batter_rae', 'sum'),
    bdry=('is_boundary', 'sum'), dots=('is_bat_dot', 'sum'), inn=('match_id', 'nunique')
).unstack(fill_value=0)
bat_agg.columns = [f"{c[0]}_{'P' if c[1] else 'N'}" for c in bat_agg.columns]
bat_agg = bat_agg.reset_index().merge(bat_outs, on='striker_id', how='left').fillna({'outs_N': 0, 'outs_P': 0})
bat_agg['last_date'] = bat_agg['striker_id'].map(last_bat)
bat_agg['ipl_p_balls'] = bat_agg['striker_id'].map(ipl_bat_p).fillna(0)

m_bat = df[df['is_pressure'] & df['is_batter_ball']].groupby(['striker_id', 'match_id']).agg(rae=('batter_rae', 'sum'), b=('is_batter_ball', 'sum')).reset_index()
bat_agg['cons_P'] = bat_agg['striker_id'].map(m_bat[m_bat['b'] >= 4].groupby('striker_id').apply(lambda g: (g['rae'] > 0).mean()).to_dict()).fillna(0.5)

bat_agg['P_SR'] = (bat_agg['runs_P'] / bat_agg['balls_P'].replace(0, np.nan)) * 100
bat_agg['P_SR_vs_Par'] = (bat_agg['rae_P'] / bat_agg['balls_P'].replace(0, np.nan)) * 100
bat_agg['N_SR_vs_Par'] = (bat_agg['rae_N'] / bat_agg['balls_N'].replace(0, np.nan)) * 100
bat_agg['Clutch_Lift'] = bat_agg['P_SR_vs_Par'] - bat_agg['N_SR_vs_Par']
bat_agg['Bdry_Pct'] = (bat_agg['bdry_P'] / bat_agg['balls_P'].replace(0, np.nan)) * 100
bat_agg['Dot_Pct'] = (bat_agg['dots_P'] / bat_agg['balls_P'].replace(0, np.nan)) * 100

elig_bat = bat_agg[(bat_agg['last_date'] >= '2025-01-01') & (bat_agg['balls_P'] >= 180) & (bat_agg['inn_P'] >= 12) & (bat_agg['ipl_p_balls'] >= 60)].copy()
C_bat = elig_bat['balls_P'].median()
elig_bat['shrunk_SR_Par'] = (elig_bat['balls_P'] * elig_bat['P_SR_vs_Par']) / (elig_bat['balls_P'] + C_bat)
elig_bat['shrunk_Lift'] = (elig_bat['balls_P'] * elig_bat['Clutch_Lift']) / (elig_bat['balls_P'] + C_bat)
elig_bat['shrunk_Cons'] = (elig_bat['inn_P'] * elig_bat['cons_P'] + 15 * elig_bat['cons_P'].mean()) / (elig_bat['inn_P'] + 15)
g_avg = elig_bat['runs_P'].sum() / elig_bat['outs_P'].sum()
elig_bat['shrunk_Avg'] = (elig_bat['runs_P'] + 5 * g_avg) / (elig_bat['outs_P'] + 5)

elig_bat['raw_ppi'] = (
    0.35 * zscore(elig_bat['shrunk_SR_Par']) + 0.20 * zscore(elig_bat['shrunk_Avg']) +
    0.15 * zscore(elig_bat['shrunk_Lift']) + 0.15 * (0.5 * zscore(elig_bat['Bdry_Pct']) - 0.5 * zscore(elig_bat['Dot_Pct'])) +
    0.15 * (0.5 * zscore(elig_bat['runs_P']) + 0.5 * zscore(elig_bat['shrunk_Cons']))
)
elig_bat['PPI'] = (50 + 15 * zscore(elig_bat['raw_ppi'])).clip(0, 100).round(2)
top10_batters = elig_bat.sort_values('PPI', ascending=False).reset_index(drop=True)
top10_batters.index += 1

# 8. Bowler Aggregation & Empirical Bayes Shrinkage
last_bowl = df.groupby('bowler_id')['match_date'].max().to_dict()
ipl_bowl_p = df[(df['comp_tier'] == 'IPL') & df['is_pressure'] & df['is_legal_ball']].groupby('bowler_id').size().to_dict()

bowl_agg = df[df['is_legal_ball']].groupby(['bowler_id', 'bowler_name', 'is_pressure']).agg(
    runs=('bowler_runs', 'sum'), balls=('is_legal_ball', 'sum'), rse=('bowler_rse', 'sum'),
    wkts=('is_bowler_wicket', 'sum'), dots=('is_bowl_dot', 'sum'), bdry=('is_boundary', 'sum'), spells=('match_id', 'nunique')
).unstack(fill_value=0)
bowl_agg.columns = [f"{c[0]}_{'P' if c[1] else 'N'}" for c in bowl_agg.columns]
bowl_agg = bowl_agg.reset_index()
bowl_agg['last_date'] = bowl_agg['bowler_id'].map(last_bowl)
bowl_agg['ipl_p_balls'] = bowl_agg['bowler_id'].map(ipl_bowl_p).fillna(0)

m_bowl = df[df['is_pressure'] & df['is_legal_ball']].groupby(['bowler_id', 'match_id']).agg(rse=('bowler_rse', 'sum'), w=('is_bowler_wicket', 'sum'), b=('is_legal_ball', 'sum')).reset_index()
bowl_agg['cons_P'] = bowl_agg['bowler_id'].map(m_bowl[m_bowl['b'] >= 4].groupby('bowler_id').apply(lambda g: ((g['rse'] > 0) | (g['w'] >= 1)).mean()).to_dict()).fillna(0.5)

bowl_agg['P_Econ'] = (bowl_agg['runs_P'] / bowl_agg['balls_P'].replace(0, np.nan)) * 6.0
bowl_agg['P_Saved_Par'] = (bowl_agg['rse_P'] / bowl_agg['balls_P'].replace(0, np.nan)) * 6.0
bowl_agg['N_Saved_Par'] = (bowl_agg['rse_N'] / bowl_agg['balls_N'].replace(0, np.nan)) * 6.0
bowl_agg['Clutch_Lift'] = bowl_agg['P_Saved_Par'] - bowl_agg['N_Saved_Par']
bowl_agg['Wkts_4Ov'] = (bowl_agg['wkts_P'] / bowl_agg['balls_P'].replace(0, np.nan)) * 24.0
bowl_agg['Dot_Pct'] = (bowl_agg['dots_P'] / bowl_agg['balls_P'].replace(0, np.nan)) * 100.0
bowl_agg['Bdry_Pct'] = (bowl_agg['bdry_P'] / bowl_agg['balls_P'].replace(0, np.nan)) * 100.0

elig_bowl = bowl_agg[(bowl_agg['last_date'] >= '2025-01-01') & (bowl_agg['balls_P'] >= 180) & (bowl_agg['spells_P'] >= 12) & (bowl_agg['ipl_p_balls'] >= 60)].copy()
C_bowl = elig_bowl['balls_P'].median()
elig_bowl['shrunk_Saved_Par'] = (elig_bowl['balls_P'] * elig_bowl['P_Saved_Par']) / (elig_bowl['balls_P'] + C_bowl)
elig_bowl['shrunk_Lift'] = (elig_bowl['balls_P'] * elig_bowl['Clutch_Lift']) / (elig_bowl['balls_P'] + C_bowl)
g_wkt = (elig_bowl['wkts_P'].sum() / elig_bowl['balls_P'].sum()) * 24.0
elig_bowl['shrunk_Wkts_4Ov'] = (elig_bowl['balls_P'] * elig_bowl['Wkts_4Ov'] + C_bowl * g_wkt) / (elig_bowl['balls_P'] + C_bowl)
elig_bowl['shrunk_Cons'] = (elig_bowl['spells_P'] * elig_bowl['cons_P'] + 15 * elig_bowl['cons_P'].mean()) / (elig_bowl['spells_P'] + 15)

elig_bowl['raw_ppi'] = (
    0.35 * zscore(elig_bowl['shrunk_Saved_Par']) + 0.25 * zscore(elig_bowl['shrunk_Wkts_4Ov']) +
    0.15 * zscore(elig_bowl['shrunk_Lift']) + 0.10 * (0.5 * zscore(elig_bowl['Dot_Pct']) - 0.5 * zscore(elig_bowl['Bdry_Pct'])) +
    0.15 * (0.5 * zscore(elig_bowl['wkts_P']) + 0.5 * zscore(elig_bowl['shrunk_Cons']))
)
elig_bowl['PPI'] = (50 + 15 * zscore(elig_bowl['raw_ppi'])).clip(0, 100).round(2)
top10_bowlers = elig_bowl.sort_values('PPI', ascending=False).reset_index(drop=True)
top10_bowlers.index += 1

print("=== TOP 10 BATTERS (SUBMIT IN UNSTOP DROPDOWN) ===")
display(top10_batters[['striker_name', 'PPI', 'runs_P', 'balls_P', 'P_SR', 'P_SR_vs_Par', 'shrunk_Avg']].head(10))

print("\n=== TOP 10 BOWLERS (SUBMIT IN UNSTOP DROPDOWN) ===")
display(top10_bowlers[['bowler_name', 'PPI', 'wkts_P', 'balls_P', 'P_Econ', 'P_Saved_Par', 'shrunk_Wkts_4Ov']].head(10))
```
